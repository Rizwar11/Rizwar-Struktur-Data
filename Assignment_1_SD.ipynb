Tugas 1: Menentukan apakah sebuah bilangan Positif, Negatif atau Nol

```python
bilangan = float(input("Masukkan sebuah bilangan: "))
if bilangan > 0:
    print("Bilangan Positif")
elif bilangan < 0:
    print("Bilangan Negatif")
else:
    print("Bilangan Nol")
```

Tugas 4: Menentukan bilangan terbesar dari 3 bilangan yang diinput user

```python
b1 = float(input("Masukkan bilangan 1: "))
b2 = float(input("Masukkan bilangan 2: "))
b3 = float(input("Masukkan bilangan 3: "))

terbesar = b1
if b2 > terbesar:
    terbesar = b2
if b3 > terbesar:
    terbesar = b3

print("Bilangan terbesar adalah:", terbesar)
```

Tugas 5: Menghitung rata-rata dari 5 nilai yang diinput user beserta kategorinya (≥ 80 Baik, 60-79 Cukup, < 60 Kurang)

```python
n1 = float(input("Masukkan nilai 1: "))
n2 = float(input("Masukkan nilai 2: "))
n3 = float(input("Masukkan nilai 3: "))
n4 = float(input("Masukkan nilai 4: "))
n5 = float(input("Masukkan nilai 5: "))

rata_rata = (n1 + n2 + n3 + n4 + n5) / 5

if rata_rata >= 80:
    kategori = "Baik"
elif rata_rata >= 60:
    kategori = "Cukup"
else:
    kategori = "Kurang"

print("Rata-rata:", rata_rata)
print("Kategori:", kategori)
```

Tugas 6: Menghitung total bayar setelah diskon (Diskon 10% jika total belanja ≥ 100.000)

```python
total_belanja = float(input("Masukkan total belanja: Rp "))

if total_belanja >= 100000:
    diskon = total_belanja * 0.10
    total_bayar = total_belanja - diskon
    print("Mendapatkan diskon 10%!")
else:
    total_bayar = total_belanja
    print("Tidak mendapatkan diskon.")

print("Total yang harus dibayar: Rp", total_bayar)
```

Tugas 7: Menampilkan tabel perkalian dari angka yang diinput user (1-10)

```python
angka = int(input("Masukkan angka: "))

for i in range(1, 11):
    hasil = angka * i
    print(f"{angka} x {i} = {hasil}")
```

Tugas 8: Menghitung faktorial dari suatu bilangan

```python
bilangan = int(input("Masukkan bilangan: "))
faktorial = 1

for i in range(1, bilangan + 1):
    faktorial *= i

print(f"{bilangan}! = {faktorial}")
```

Tugas 9: Menentukan apakah suatu bilangan termasuk bilangan prima atau bukan

```python
bilangan = int(input("Masukkan bilangan: "))

if bilangan > 1:
    for i in range(2, int(bilangan/2) + 1):
        if (bilangan % i) == 0:
            print("Bukan bilangan prima")
            break
    else:
        print("Bilangan prima")
else:
    print("Bukan bilangan prima")
```

Tugas 10: Membuat program login (Username = "admin", Password = "123", Maksimal percobaan = 3 kali. Jika gagal 3 kali tampilkan "Akses ditolak")

```python
percobaan = 3

while percobaan > 0:
    username = input("Masukkan Username: ")
    password = input("Masukkan Password: ")

    if username == "admin" and password == "123":
        print("Login Berhasil!")
        break
    else:
        percobaan -= 1
        print(f"Username atau Password salah! Sisa percobaan: {percobaan}")

if percobaan == 0:
    print("Akses ditolak")
```
